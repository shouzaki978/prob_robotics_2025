# 1次元カルマンフィルタ


## 状態ベクトルの定義

ロボットの状態を次のように定義する．

$$
\mathbf{x}_t =
\begin{pmatrix}
x_t \\
v_t
\end{pmatrix}
$$

ここで，$x_t$ は位置，$v_t$ は速度を表す．

---

## 状態方程式（運動モデル）

ロボットの運動は，加速度入力を含むモデルで表される．

$$
\mathbf{x}_t = \mathbf{A} \mathbf{x}_{t-1} + \mathbf{B} u_t + \mathbf{G} w_t
$$

ここで,
- $ u_t $ : 既知の制御入力(加速度)
- $ \mathbf{G} w_t $ : ノイズ

であり，$w_t$は次の正規分布に従う確率変数である.

$$
w_t \sim \mathcal{N}(0, \sigma_a^2)
$$

また,定数行列 $ \mathbf{A}, \mathbf{B}, \mathbf{G},$ は次のように与えられる.
$$
\mathbf{A} =
\begin{pmatrix}
1 & \Delta t \\
0 & 1
\end{pmatrix},
\quad
\mathbf{B} =
\begin{pmatrix}
\frac{\Delta t^2}{2} \\
\Delta t
\end{pmatrix},
\quad
\mathbf{G} = \mathbf{B}
$$

---

## 観測方程式

観測は位置のみとして, 次のようなモデルを用いる

$$
z_t = \mathbf{H} \mathbf{x}_t + v_t 
$$

$$
\mathbf{H} =
\begin{pmatrix}
1 & 0
\end{pmatrix}
$$

ここで,
- $ z_t $ : 観測値
- $ v_t $ : 観測ノイズ
- $ \mathbf{H} $ は位置 $ \mathbf{x}_t $  で得られるかをモデル化した行列

であり, $ v_t $ は次の正規分布に従うノイズである.

$$
v_t \sim \mathcal{N}(0, \sigma_z^2)
$$
